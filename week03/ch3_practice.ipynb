## 1. TensorFlow: scalar derivatives
For f(x, y) = x²y + xy + 3y at x = 2, y = 1, the expected derivatives are 5 and 9.

```python
import tensorflow as tf
import numpy as np

x = tf.Variable(2.0)
y = tf.Variable(1.0)
with tf.GradientTape() as tape:
    f = x**2 * y + x*y + 3*y
dx, dy = tape.gradient(f, [x, y])
print("df/dx:", dx.numpy())
print("df/dy:", dy.numpy())
np.testing.assert_allclose([dx.numpy(), dy.numpy()], [5, 9])
```

## 2. TensorFlow: linear expression
Compute gradients of the sum of the two outputs of xW + b.

```python
x = tf.constant([[1.0, 4.0, 3.0]])
W = tf.Variable(tf.random.uniform((3, 2)))
b = tf.Variable(tf.zeros((2,)))
with tf.GradientTape() as tape:
    output = tf.matmul(x, W) + b
    total = tf.reduce_sum(output)
dW, db = tape.gradient(total, [W, b])
print("Gradient with respect to W:")
print(dW.numpy())
print("Gradient with respect to b:")
print(db.numpy())
np.testing.assert_allclose(dW.numpy(), [[1, 1], [4, 4], [3, 3]])
np.testing.assert_allclose(db.numpy(), [1, 1])
```

## 3. TensorFlow: cubic expression
Cube each output and sum them before differentiation. For z = xW + b, db = 3z² and dW is the outer product of x and db.

```python
with tf.GradientTape() as tape:
    output = tf.matmul(x, W) + b
    total = tf.reduce_sum(tf.pow(output, 3))
dW, db = tape.gradient(total, [W, b])
print("Gradient with respect to W (cubic):")
print(dW.numpy())
print("Gradient with respect to b (cubic):")
print(db.numpy())
expected_db = 3 * output.numpy()[0]**2
np.testing.assert_allclose(db.numpy(), expected_db, rtol=1e-5)
np.testing.assert_allclose(dW.numpy(), x.numpy().T * expected_db, rtol=1e-5)
```

## 4. PyTorch: scalar derivatives
Repeat the first exercise with PyTorch.

```python
import torch

x = torch.tensor(2.0, requires_grad=True)
y = torch.tensor(1.0, requires_grad=True)
f = x**2 * y + x*y + 3*y
f.backward()
print("df/dx:", x.grad.item())
print("df/dy:", y.grad.item())
assert (x.grad.item(), y.grad.item()) == (5.0, 9.0)
```

## Conclusion
TensorFlow and PyTorch give the same scalar derivatives: 5 and 9. Computing gradients does not itself update parameters. The linear and cubic exercises use the sum of the two outputs as the scalar differentiation target.
